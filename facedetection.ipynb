```python
import cv2 as cv2
```

```python
import mediapipe as mp
```

```python
import numpy as np
```

```python


# Initialize MediaPipe Face Mesh
mp_face_mesh = mp.solutions.face_mesh

# Load the image
image = cv2.imread('Tutorial9\\image.jpg')

# Convert the image to RGB (MediaPipe expects RGB images)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Initialize the face mesh model
with mp_face_mesh.FaceMesh(static_image_mode=True, max_num_faces=1, refine_landmarks=True, min_detection_confidence=0.5) as face_mesh:
    results = face_mesh.process(image_rgb)

    if results.multi_face_landmarks:
        for face_landmarks in results.multi_face_landmarks:
            h, w, _ = image.shape
            
            # Get landmarks for the key facial features: eyebrows, eyes, nose, lips
            left_eyebrow_landmark = face_landmarks.landmark[70]  # Left eyebrow top
            right_eyebrow_landmark = face_landmarks.landmark[300]  # Right eyebrow top
            left_eye_landmark = face_landmarks.landmark[33]  # Left eye corner
            right_eye_landmark = face_landmarks.landmark[263]  # Right eye corner
            nose_tip_landmark = face_landmarks.landmark[1]  # Nose tip
            top_lip_landmark = face_landmarks.landmark[13]  # Upper lip center
            bottom_lip_landmark = face_landmarks.landmark[14]  # Lower lip center

            # Calculate bounding box
            x_min = int(min(left_eyebrow_landmark.x, right_eyebrow_landmark.x, left_eye_landmark.x, right_eye_landmark.x, nose_tip_landmark.x, top_lip_landmark.x, bottom_lip_landmark.x) * w)
            y_min = int(min(left_eyebrow_landmark.y, right_eyebrow_landmark.y, left_eye_landmark.y, right_eye_landmark.y, nose_tip_landmark.y, top_lip_landmark.y, bottom_lip_landmark.y) * h)
            x_max = int(max(left_eyebrow_landmark.x, right_eyebrow_landmark.x, left_eye_landmark.x, right_eye_landmark.x, nose_tip_landmark.x, top_lip_landmark.x, bottom_lip_landmark.x) * w)
            y_max = int(max(left_eyebrow_landmark.y, right_eyebrow_landmark.y, left_eye_landmark.y, right_eye_landmark.y, nose_tip_landmark.y, top_lip_landmark.y, bottom_lip_landmark.y) * h)

            # Add some margin to ensure the entire face is captured
            margin = 20
            x_min = max(0, x_min - margin)
            y_min = max(0, y_min - margin)
            x_max = min(w, x_max + margin)
            y_max = min(h, y_max + margin)

            # Extract the region containing all key features (from eyebrows to lips)
            face_crop = image[y_min:y_max, x_min:x_max]

            # Stretch the cropped image horizontally to 224x224 (ignore aspect ratio)
            face_crop_resized = cv2.resize(face_crop, (224, 224), interpolation=cv2.INTER_LINEAR)

            # Display the stretched image
            cv2.imshow('224x224 Stretched Face Crop', face_crop_resized)
            cv2.waitKey(0)
            cv2.destroyAllWindows()
```
